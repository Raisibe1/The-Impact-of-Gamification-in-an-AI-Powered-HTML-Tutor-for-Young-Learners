```python
import pandas as pd
import os
```

```python
# 1. Load the final syllabus
syllabus_df = pd.read_csv('../data/reference/html_syllabus.csv')
print("✅ Final HTML Syllabus loaded successfully!")
display(syllabus_df)
```

```python
# 2. Quick statistics
print(f"\n📊 Total topics: {len(syllabus_df)}")
print(f"\n📚 Grade distribution:")
print(syllabus_df['grade_level'].value_counts().sort_index())
```

```python
print(f"\n⭐ Difficulty distribution:")
print(syllabus_df['difficulty'].value_counts().sort_index())
```

```python
# 3. Show which sources were used
print("\n📖 Source evidence for each topic:")
for _, row in syllabus_df.iterrows():
    print(f"  {row['topic_id']} - {row['topic_name']}: {row['source_evidence']}")
```
